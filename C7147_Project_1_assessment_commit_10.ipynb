## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors

### 🎯 Aims:

- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Invalid reads should not crash your program.

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """With valid data, the function will find bases in the sequence that are difference from the base and will return the position of the mismatch with the incorrect bases."""
    # Write your code here.
    mutations = []
    # sets mutations as a variable of empty list
    for base in range(len(reference)):
      #loops the following "if" statement for as many characters in the "reference" string
      if sequence[base] != reference[base]:
        # checks if the to see if each value (base) in the sequence does not match each base in the reference
        mutations.append({"position": base + 1, "reference_base": reference[base],"read_base": sequence[base]})
        # sets a dictionary for each loop cycle to have an individual position each and to record the base value for reference and sequence
        # this is done on the condition that there is a mismatch before sequence and reference
    return (mutations)
    # this returns the values to be used outside of the function
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """This functions uses validate_read function to validate the sequences in the list of dictionarys (reads) and using find_mutations function to find mismatch bases betweeen the sequences and reference. \n Then a summary of both with additional detail is returned"""
    # Write your code here
    analysis = []
    # sets summary for this function as an empty list to be used as dictionary later
    for sequences in reads:
      # loops following code including dictionary append until the end of rows in reads
      results = validate_read(sequences["sequence"], reference)
      # uses a previously defined function to validate reads
      if results["valid"] == True:
        # checks for a value from the validate_read function is true
        status = "valid"
        # sets a new string variable of status as "valid"
        mutations = find_mutations(sequences["sequence"].upper(), reference)
        # and then moves onto find_mutations function to check when sequence does not match reference and store it
        # upper() is used as find_mutations does not check for and does not accept lowercase since lowercase does pass in validate_read check
      else:
        status = "invalid"
        # status is invalid when results["valid"] is not true
        mutations = []
        # sets mutations as an empty list when the sequence is invalid and cannot be entered into find_mutations function
      analysis.append({"read_id": sequences["read_id"],
                    "sample": sequences["sample"],
                    "group": sequences["group"],
                    "sequence": results["sequence"],
                    "status": status,
                    "errors": results["errors"],
                    "mutations": mutations})
      # this puts all of our needed values into a dictionary for each sample as it is include in the loop.
    return analysis
    # this outputs our summary dictionary outside the function to be used
````

````python
# Write your code here.
def summarise_samples(analyse_reads):

  summaries = {}
  # Creates our dictionary to store our original base for output

  substitutions =["A>T","A>G","A>C",
                 "T>A","T>G","T>C",
                 "C>T","C>G","C>A",
                 "G>A","G>T","G>C"]
  # creates a list of the possible substitions

  for read in analyse_reads:
    groups = read["group"]
    samples = read["sample"]
    if samples not in summaries:
      summaries[samples] = {"samples": samples,
                            "group": groups,
                            "valid_reads": 0,
                            "invalid_reads": 0,
                            "mismatch_reads": 0,
                            "mismatch_percentage": 0,
                            "mismatch_events": 0,
                            "substitution_counts": {},
                            "position_counts": {},
                            "top_positions": []}
  # assigns groups and samples variables from the data input
  # if there are no samples, the output is a summary of empty values

    summary = summaries[samples]
  # is a copy that does not interfere with the orignal summaries
  # and can be altered per sample

    if read["status"] == "invalid":
      summary["invalid_reads"] += 1
  # this is our count for invalid samples using status from input data

    elif read["status"] == "valid":
      summary["valid_reads"] += 1
  # our count for valid samples

      mutations = read["mutations"]
  # makes input mutations functional inside here

      if len(mutations) > 0:
        summary["mismatch_reads"] += 1
        summary["mismatch_events"] += len(mutations)
  # tallies that reads with mismatches
  # and how many mismatches in all reads occur

        for mutation in mutations:
          position = mutation["position"]
  # identifies the position of each mismatch

          substitution = (mutation["reference_base"] + ">" + mutation["read_base"])
          if substitution in substitutions:
            summary["substitution_counts"][substitution] = (summary["substitution_counts"].get(substitution, 0) + 1)

  # identifies which base nucleotides were substituted out for what
  # and counting based on our list of substitutions defined earlier in this function

          summary["position_counts"][position] = (summary["position_counts"].get(position,0) + 1)

  # counts how many times a mismatch occurs in a position

  results={}
  # sets up an output variable
  for samples, summary in summaries.items():
    if summary["valid_reads"] == 0:
      summary["mismatch_percentage"] = None
    else:
      summary["mismatch_percentage"] = (summary["mismatch_reads"]/summary["valid_reads"]) * 100
  # as it is a new loop that is set valid reads is 0 then the mismatch percentage is none as it cannot be calculated
  # otherwise (the sample is valid) and the mismatch_percentage is calculated from how many reads in a sample have at least one mismatch as a percentage

    top_positions = sorted(summary["position_counts"].items(), key = lambda item: (-item[1], item[0]))[:3]
    summary["top_positions"] = [position for position, count in top_positions]

  # this sorted and displays the top 3 positions with the most mutations from highest to lowest

  groups = {}
  for sample, summary in summaries.items():
    group = summary["group"]
  # separates the samples by groups

    if group not in groups:
      groups[group] = []
    groups[group].append(sample)

  # adds a group based on different samples if there is not one in the dataset allowing for more groups to summarised
  # than initially defined or just multiple as needed.

    results[sample] = {"samples": samples,
                       "group": summary["group"],
                      "valid_reads": summary["valid_reads"],
                      "invalid_reads": summary["invalid_reads"],
                      "mismatch_reads": summary["mismatch_reads"],
                      "mismatch_percentage": summary["mismatch_percentage"],
                      "mismatch_events": summary["mismatch_events"],
                      "substitution_counts": summary["substitution_counts"],
                      "position_counts": summary["position_counts"],
                      "top_positions": summary["top_positions"]}
  # formats my values from each summary per sample to be a dictionary stored in results to be returned and output
  for sample, summary in summaries.items():
    print(f'"group": "{summary["group"]}",')
    print(f'"valid_reads": {summary["valid_reads"]},')
    print(f'"invalid_reads": {summary["invalid_reads"]},')
    print(f'"mismatch_reads": {summary["mismatch_reads"]},')
    print(f'"mismatch_percentage": {summary["mismatch_percentage"]},')
    print(f'"mismatch_events": {summary["mismatch_events"]},')

    print('"position_counts": {')
    for position, count in summary["position_counts"].items():
      print(f'    {position}: {count},')
    print('},')

    print('"substitution_counts": {')
    for substitution, count in summary["substitution_counts"].items():
      print(f'    "{substitution}": {count},')
    print('},')

    print(f'"top_positions": {summary["top_positions"]}')
  # printing of the table

  return(results)


````

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````
